```python
import pandas as pd
import numpy as np
from math import log2

# Dataset
data = pd.DataFrame({
    'Outlook':['Sunny','Sunny','Overcast','Rain'],
    'Temp':['Hot','Hot','Hot','Mild'],
    'Play':['No','No','Yes','Yes']
})

# Entropy
def entropy(y):
    p = y.value_counts(normalize=True)
    return -sum(p * np.log2(p))

# Info Gain
def gain(data, col, target):
    total = entropy(data[target])
    vals = data[col].unique()
    weighted = sum((len(data[data[col]==v])/len(data)) *
                   entropy(data[data[col]==v][target]) for v in vals)
    return total - weighted

# Best feature
features = ['Outlook','Temp']
best = max(features, key=lambda x: gain(data, x, 'Play'))

print("Best Feature:", best)
```
